# Theorie

Volgens [Newton's wet van afkoeling](https://nl.wikipedia.org/wiki/Afkoelingswet_van_Newton) is de snelheid waarmee een voorwerp afkoelt evenredig met het verschil in de temperatuur van het voorwerp ($T$) en de omgeving ($T_0$). We kunnen dit schrijven als:

$$
    \dot{Q} = -hA(T(t) - T_0),
$$

waarin 

- $\dot{Q}$ de warmtestroom in $\mathrm{W}$,
- $A$ het oppervlak waardoor koeling optreedt in $\mathrm{m}^2$,
- $h$ de warmteoverdrachtscoëfficiënt in $\mathrm{W/(m^2 K)}$.

Dit levert de differentiaalvergelijking

$$
    C\dot{T} = -hA(T(t) - T_0),
$$

met $C$ de warmtecapaciteit in $\mathrm{J/kg}$. Herschrijven met $\tau = \frac{C}{hA}$ levert:

$$
    -\tau\dot{T} = T(t) - T_0,
$$ (eq:dv1)

met als oplossing:

$$
    T(t) - T_0 = (T(0) - T_0)\text{e}^{-t/\tau}.
$$ (eq:dv1_opl)

We kunnen hieruit dus concluderen dat $\tau$ de karakteristieke tijdsduur is waarin de temperatuur van de buis een factor $\text{e}$ verlaagd ten opzichte van de omgevingstemperatuur. 

```{exercise} Oplossing DV controleren 
:label: ex-dvopl
Controleer bovenstaande door vergelijking [](#eq:dv1_opl) te substitueren in [](#eq:dv1).
```

````{solution} ex-dvopl
−τ dT/dt = T(t) − T₀        (3) dit is vergelijking drie. 
T(t) − T₀ = (T(0) − T₀) · e^(−t/τ)     (4) waarin de subsitutie moet. 
T(t) = T₀ + (T(0) − T₀) · e^(−t/τ), hierin T nul naar de andere kant brengen om T(t) expliciet te krijgen.  door de afgeleide te berekenen kan de originele vergelijking  weer terug gevonden worden. Afgeleide geeft: 
dT/dt = d/dt [ T₀ + (T(0) − T₀) · e^(−t/τ) ]

dT/dt = (T(0) − T₀) · d/dt [ e^(−t/τ) ]

dT/dt = (T(0) − T₀) · (−1/τ) · e^(−t/τ)

dT/dt = −(T(0) − T₀)/τ · e^(−t/τ)
Als vervolgens wordt vermenigvuldig met de karakteristieke tijd dan wordt vergelijking terug gevonden: −τ · dT/dt
= −τ · [ −(T(0) − T₀)/τ · e^(−t/τ) ]

= (T(0) − T₀) · e^(−t/τ)
En dus klopt de vergelijking −τ dT/dt = T(t) − T₀

````

We zijn hier voor het gemak uitgegaan van een $h$ die onafhankelijk is van de temperatuur. We weten echter dat warmtetransport door straling niet lineair gaat, maar als 

$$
    \dot{Q}_s = \epsilon \sigma A (T^4 - T_0^4).
$$ 

Voor kleine temperatuurverschillen ($\Delta T = T - T_0$) is dit te vereenvoudigen tot

$$
    \dot{Q}_s = \epsilon \sigma A ((T_0+\Delta T)^4 - T_0^4) \approx \epsilon \sigma 4A T_0^3 \Delta T.
$$

Zolang $\Delta T$ dus relatief klein is ten opzichte van $T_0$, kunnen we $h$ dus inderdaad als een constante beschouwen.

`````python
# Hier de data en de analyse

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Afmetingen buis


hoogte = 0.10181 # in meter
diameter1 = 0.0500 # diameter buitenkant van de buis, in meter
dikte = 0.00170 # in meter
diameter2 = diameter1 - 2 * dikte # diameter binnenkant van de buis

buitenoppervlak = hoogte * diameter1 * np.pi
volume1 = np.pi * (0.5*diameter1)**2 * hoogte
volume2 = np.pi * (0.5*diameter2)**2 * hoogte

volume = volume1-volume2
dichtheid = 8.73*10**3
gewicht = volume * dichtheid
C = 3.8*10**2

omgtemp = 21.1
starttemp1 = 52.5
starttemp2 = 54.0

def exp_func1(t, A, tau, T_omg):
    # A is verschiltemperatuur met omgeving aan start
    # tau is de karakteristieke tijd voor de koeling
    # T_omg is de omgevingstemperatuur
    return (A * np.exp(-t/tau) + T_omg)  

def exp_func2(t, A, tau, T_omg):
    # A is verschiltemperatuur met omgeving aan start
    # tau is de karakteristieke tijd voor de koeling
    # T_omg is de omgevingstemperatuur
    return (A * np.exp(-t/tau) + T_omg)  

warmtecapaciteit = gewicht * C # bepaal de warmtecapaciteit in J/K

# Metingen van de buis.
times = np.array([0, 30, 42, 52, 60, 75, 85, 95, 102, 110, 120, 130, 140, 145, 153, 160, 165, 175, 182, 188, 200, 210, 220, 225, 235, 251, 260]) # in seconden
temps = np.array([52.0, 51.5, 49.8, 48.7, 48.2, 47.6, 47.0, 46.0, 45.4, 45.0, 44.5, 43.9, 43.4, 42.7, 42.3, 41.8, 41.3, 40.7, 40.2, 39.8, 39.3, 38.8, 38.3, 38.4, 37.6, 37.1, 36.6]) # in celsius

# Metingen van de buis met dop erin
times2 = np.array([0, 20, 35, 45, 50, 58, 63, 70, 75, 84, 95, 100, 108, 115, 125, 130, 137, 150, 158, 165, 175, 183, 195, 210, 215, 220, 230, 235, 240, 245, 255, 265, 270, 275, 280, 290, 300, 305, 315])
temps2 = np.array([54.0, 53.6, 53.0, 52.5, 51.8, 51.3, 50.8, 49.8, 49.2, 48.8, 48.3, 47.8, 47.3, 46.8, 46.3, 46.3, 45.8, 45.3, 44.8, 44.3, 43.8, 43.3, 42.8, 42.4, 41.9, 41.9, 41.4, 41.4, 40.9, 40.9, 40.4, 40.0, 40.0, 39.9, 39.9, 39.4, 38.9, 38.9, 38.4])

# pas beginwaardes aan naar schatting
# Het aantal maxfev moet wellicht hoger voor goede convergentie van de waarde
popt, pcov = curve_fit(exp_func1, times, temps, p0=[30, 1000, 20], maxfev=10000)
popt2, pcov2 = curve_fit(exp_func2, times2, temps2, p0=[30, 1000, 20], maxfev=10000)

A_exp, tau_exp, T_omg_exp = popt
A_exp2, tau_exp2, T_omg_exp2 = popt2

y_fit = exp_func1(times, *popt)
y_fit2 = exp_func2(times2, *popt2)


plt.figure()
plt.xlabel('Time [s]')
plt.ylabel('Temperature [K]')

plt.plot(times, temps, 'bo', label='meting')

plt.plot(times, y_fit, 'r-', 
         label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp, tau_exp, T_omg_exp))

plt.legend()
plt.show()

plt.figure()
plt.xlabel('Time [s]')
plt.ylabel('Temperature [K]')

plt.plot(times2, temps2, 'bo', label='meting met dop')
plt.plot(times2, y_fit2, 'r-', 
         label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp2, tau_exp2, T_omg_exp2))

plt.legend()

plt.show()

h_exp = (warmtecapaciteit) / (tau_exp * buitenoppervlak)
h_exp2 = (warmtecapaciteit) / (tau_exp2 * buitenoppervlak)

print("Karakteristieke tijd zonder dop: ", tau_exp ,'s')
print("Karakteristieke tijd met dop: ", tau_exp2 ,'s')
print("warmteoverdachtscoëfficiënt zonder dop: ", h_exp, "W/m^2K")
print("warmteoverdachtscoëfficiënt met dop: ", h_exp2,  "W/m^2K" )
print("geleiding zogt voor verdeling van de warmte over de buis, maar berperkt niet warmte verlies")
print("Het aandeel van afkoeling zal dus komen door convenctie en straling")
print("Komt doordat de buis omringd wordt door lucht en er een temperatuur verschil is tussen de buis en de omgeving")

 # warmteoverdrachtscoëfficiënt in W/m^2 K
# Sla figuren op met  
# 
# plt.savefig("Figuren/naam.png", dpi=450)

`````

Antwoord op de vraag: Welk deel van de warmteoverdrachtscoëfficiënt verwacht je dat gegeven is door de geleiding, straling en convectie? Onderbouw je redenering.

warmte overdracht is de som van de warmte overdacht door geleiding, convectie en straling. geleiding zorgt ervoor dat de warmte binnen de cilinder wordt verdeelt, maar zorgt niet voor warmteoverdacht naar de omgeving. 

De warmte overdracht met straling heeft vooral te maken met de oppervlakte van het objeect emissiviteit en de verschil in temperatuur. Aangezien dat het verschil in temperatuur relatief klein is en straling niet beïnvloed wordt door de orientatie van een buis en de dop, zal waarschijnlijk de grootste factor komen door convectie, waar dit wel het geval is.
