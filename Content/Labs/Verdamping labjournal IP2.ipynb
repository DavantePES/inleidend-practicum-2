# Is het verdamping?

In dit experiment is er een verwarmingselement in een met water gevulde maatbeker gestopt. Elke minuut is de temperatuur van het water gemeten. Deze metingen zijn opgeslagen in [tempmetingen.csv](tempmetingen.csv).

Verder is gegeven dat: 
$m_{maatbeker} = 820.8 \mathrm{gr}$  
$m_{maatbeker + water, start} = 1292.9 \mathrm{gr}$  
$m_{maatbeker + water, eind} = 1274.9 \mathrm{gr}$

Het moge duidelijk zijn dat er water is 'verdwenen'. Ook is de eindtemperatuur van het water niet gelijk aan een proces waarbij verdamping en eventuele warmteverliezen niet meegenomen worden.

3. Geef aanbevelingen voor een verbetering van het experiment om (nog) beter het thermodynamische proces te kunnen onderzoeken en beschrijven.

````python
import numpy as np
import matplotlib.pyplot as plt
c = r"C:\Users\roelv\Documents\IP2\tempmetingen.txt"
data = np.loadtxt(c, delimiter=',', skiprows=1)
t = data[: ,0]
T = data[: , 1]
# fysische parameters
m = 0.4721        # kg
c_w = 4180        # J/(kg K)
P = 100           # W (pas aan indien nodig)

# tijd in seconden!
t_sec = t * 60

# begintemperatuur uit eerste meting
T0 = T[0]

# model
T_model = T0 + (P / (m * c_w)) * t_sec
plt.plot(t, T, 'k.', label="data")
plt.plot(t, T_model, 'b--', label='verwachte trendlijn (model)')
plt.xlabel('$t$ (s)')
plt.ylabel('$T$ (°C)')
plt.xlim(min(t), max(t))
````

````python

# massa's (in gram)
m_maatbeker = 820.8
m_start = 1292.9
m_eind = 1274.9

# temperaturen (uit metingen)
T_start = 21.2      #In  °C
T_eind = 49.4       # In °C
c_water = 4180      # In J/(kg*K)
L_v = 2.26e6        # In J/kg (verdampingswarmte water)

# Hierbij is de aaname gemaakt dat het
# verwarmingselement ongeveer 100 W gebruikt

P = 100             # W 
t_min = 22          # minuten
t_sec = t_min * 60  # seconden

# de begin massa van het water
m_water = (m_start - m_maatbeker) / 1000      # kg

# De massa van het wat dat is verdampt
m_verdamp = (m_start - m_eind) / 1000          # kg

# temperatuurverandering
delta_T = T_eind - T_start                     # K

# toegevoegde energie
Q_toegevoegd = P * t_sec                       # J

# energie voor de opwarming van het water
Q_opwarming = m_water * c_water * delta_T      # J

# energie die het  verdampte water bevat
Q_verdamping = m_verdamp * L_v                 # J

# De totale energie
Q_nodig = Q_opwarming + Q_verdamping            # J


print("Totale benodigde energie (J) met een verwarmings element van 100W:", Q_nodig, "J")

````

De data volgt niet de trendlijn, omdat er water verloren gaat door verdamping. De verdamping kost veel energie, waardoor de temperatuur niet zal stijgen. Dit betekent dat de trendlijn zoals die is geplot boven de datapunten ligt. Een aanbeveling om het experiment beter uitvoeren zou kunnen zijn dat de beker beter wordt afgsloten en geïsoleerd, waardoor er minder warmte verdwijnt die niet voor temperatuurstijgingen gebruikt kunnen worden. Daarnaast is het meten van het vermogen van het verwarmingselement belangrijk, zodat een betere schatting kan worden gemaakt hoeveel energie aan het water wordt overgedragen en er verdwijnt.
